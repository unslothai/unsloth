# Unsloth Studio on Kaggle

Enable a **GPU accelerator** (Settings → Accelerator → GPU T4 / P100).

This notebook starts Studio through a **private Cloudflare named tunnel** (not a public `*.trycloudflare.com` quick tunnel) and reuses credentials from **Add-ons → Secrets**:

| Secret | Purpose |
|---|---|
| `CLOUDFLARE_TUNNEL_TOKEN` | Cloudflare tunnel token (`cloudflared tunnel run --token`) |
| `CLOUDFLARE_TUNNEL_URL` | Stable hostname, e.g. `https://studio.example.com` |
| `UNSLOTH_STUDIO_ADMIN_PASSWORD` | Admin login (persists across ephemeral Kaggle disks) |
| `UNSLOTH_STUDIO_AUTH_TOKEN` | API key, must start with `sk-unsloth-` |

Point `UNSLOTH_GIT` at **your fork** so Kaggle runs these Studio patches even if upstream does not merge them.
